```python
#defining a class blueprint
class Car:
    #making a constructor
    def __init__(self,model,make,color,fuel,speed):
        #these are the attributes (what it has )
        self.model=model
        self.color=color
        self.make=make
        self.fuel=fuel
        self.__speed=speed #private variable (hidden)
        
    #defining methods (what it does )
    def drive(self):
        self.speed=60
        print(f"{self.model} of {self.color} of the year {self.make} is on the road with speed {self.__speed}")

    def stop(self):
        self.speed=0
        print(f"{self.model} of {self.color} of the year {self.make} has stopped with speed {self.__speed}")

#creating objects (real instances made from the class)
Car1=Car("audi", 2020, "gray", "petrol", 30)
Car2=Car("BMW",2019,"white","diesel",40)

Car1.drive()
Car2.stop()       
```

## **Encapsulation**  
Hiding details inside a class and exposing only what is necessary.
(Like you drive a car without knowing how the engine works).

```python
#defining a class blueprint
class Car:
    #making a constructor
    def __init__(self,model,speed):
        #these are the attributes (what it has )
        self.model=model
        self.__speed=40 #private variable (hidden)
        
    #public method to increase speed
    def accelerate(self, amount):
        self.__speed=self.__speed+amount
        print(f"the speed is {self.__speed}")

    #public method to get speed
    def get_speed(self):
        print(f"{self.__speed}")
        

#creating objects (real instances made from the class)
Car1=Car("audi", 30)
Car2=Car("BMW",40)

Car1.accelerate(30)
Car2.get_speed()       
```

## **Inheritance**  
A class can inherit properties/behaviors from another class.
after making a class of car we can make another class as electricCar. since the overall bone structure will be the same(it is a vehicle) only the attribute fuel system will change from petrol/diesel to battery.

### Super()  
super() is used in inheritance to call methods from the parent class inside the child class.

```python
#making a Car CLASS
class Car:
    #constructor
    def __init__(self, model,color,make,speed):
        self.model=model
        self.color=color
        self.make=make
        self.speed=speed 

# making another class that will use the properties of the above class
class ElectricCar(Car):
    #only battery added extra here 
    def __init__(self,model,color,make,speed, battery):
        super().__init__(model,color,make,speed)
        self.battery=battery

    def run(self):
        print(f"The car {self.model} is running on {self.battery}% battery")

ecar=ElectricCar("Tesla", "silver", 2024, "130", 78)
ecar.run()
```

## **Polymorphism**  
Same method name, different behavior depending on the object.  
(Like start() means different for a petrol car vs. electric car).  
like grade() could mean different grades for different marks of students  
In the example below we have used same method called as start but different outputs depending upon the type of the car.

```python
#making a Car CLASS
class Car:
    #constructor
    def __init__(self, model,color,make,speed):
        self.model=model
        self.color=color
        self.make=make
        self.speed=speed 

    def start(self):
        print(f"the car runs on fuel")

# making another class that will use the properties of the above class
class ElectricCar(Car):
    #only battery added extra here 
    def __init__(self,model,color,make,speed, battery):
        super().__init__(model,color,make,speed)
        self.battery=battery

    def start(self):
        print("the car runs on battery")

Car1=Car("audi", "gray", 2020, 30)
ecar=ElectricCar("Tesla", "silver", 2024, "130", 78)

ecar.start()
Car1.start()
```

## File System  
the overall structure is like   
open("filename", "mode")  
Common Modes:  
"r" → Read (default). File must exist.  
"w" → Write. Creates new file or overwrites existing file.  
"a" → Append. Adds data at the end of file.  
"rb" / "wb" → Read/Write in binary mode.

#### Read a file

```python
# Open file in read mode
f = open("myfile.txt", "r")

#read content
content=f.read()
print(content)

# Close file
f.close()
```

#### Write to a file

```python
f = open("myfile.txt","w")
f.write("Hello, this is Python!\n")
f.write("Writing to a file is easy.\n")
f.write("writing inside a file using write function\n")
f.close()
```

```python
f = open("myfile.txt", "a")   # "a" adds to file
f.write("\nThis line is appended.")
f.close()
```

```python
with open("myfile.txt", "r") as f:
    content=f.read()
    print(content)
```
